```python
from pathlib import Path
from dnd2024_character_json import validate, generate_sheets

# Point to your local workspace 'data' folder
workspace = Path.cwd() / "data"

print(f"Target Workspace: {workspace}")
print(f"Characters Folder exists: {(workspace / 'characters').exists()}")
print(f"Output Folder exists: {(workspace / 'output').exists()}")
```

```python
print("--- Running Character Validation ---")
try:
    validate.main(workspace_dir=workspace)
except SystemExit as e:
    # Jupyter catches sys.exit() calls from scripts. 
    # Exit code 0 means success, anything else means validation errors found.
    if e.code == 0:
        print("\nSUCCESS: All character files are valid!")
    else:
        print(f"\nFAILURE: Validation failed with exit code {e.code}.")
```

```python
print("--- Generating Character Sheets ---")
try:
    generate_sheets.main(workspace_dir=workspace)
    print("\nSUCCESS: PDF character sheets generated successfully!")
except Exception as e:
    print(f"\nError during generation: {e}")
```
